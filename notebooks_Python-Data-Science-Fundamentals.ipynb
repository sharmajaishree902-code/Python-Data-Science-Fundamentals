```python
#Check Whether a Number Is Even or Odd
def check_number(number):

    if number % 2 == 0:
        return "Even Number"

    else:
        return "Odd Number"


print(check_number(15))
print(check_number(20))
```

```python
#Calculate Factorial
def factorial(number):

    result = 1

    for i in range(1, number + 1):
        result = result * i

    return result


print("Factorial:", factorial(5))
```

```python
#Mini Project 1: Student Marks Calculator
def calculate_average(marks):

    total = sum(marks)

    average = total / len(marks)

    return total, average


marks = [85, 90, 78, 92, 88]

total, average = calculate_average(marks)

print("Total Marks:", total)
print("Average Marks:", average)

if average >= 90:
    print("Grade: A")

elif average >= 75:
    print("Grade: B")

elif average >= 60:
    print("Grade: C")

else:
    print("Grade: D")
```

```python
#Mini Project 2: Number Guessing Game
import random

secret_number = random.randint(1, 20)

print("Welcome to Number Guessing Game!")

while True:

    guess = int(input("Guess a number between 1 and 20: "))

    if guess == secret_number:
        print("Congratulations! You guessed correctly.")
        break

    elif guess < secret_number:
        print("Try a higher number!")

    else:
        print("Try a lower number!")
```

```python
#Mini Project 3: Simple Student Management System
students = []

def add_student(name, course):

    student = {
        "name": name,
        "course": course
    }

    students.append(student)

    print("Student Added Successfully!")


def display_students():

    for student in students:

        print("Name:", student["name"])
        print("Course:", student["course"])
        print("-------------------")


add_student("Jaishree", "BCA Data Science")
add_student("Priya", "BCA")

display_students()
```
